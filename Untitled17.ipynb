```python
import requests
import time


def search_whit_artworks(title_contains=None):
    """
    Search for artworks by keyword.

    Args:
        query: Search term (e.g., "impressionism", "cats", "van gogh")

    Returns:
        List of artwork IDs
    """
    print(f"Searching for artworks about '{title_contains}'...")

    base_url = "https://whitney.org/api"
    search_url = f"{base_url}/artworks"
    params = {"q[title_cont]": title_contains}

    try:
        response = requests.get(search_url, params=params)

        if response.status_code == 200:
            data = response.json()

            artworks_data = data.get("data", [])
            object_ids = [artwork['id'] for artwork in artworks_data]
            total = len(object_ids)

            print(f"Found {total} artworks")
            return object_ids
        else:
            print(f"Search failed with status {response.status_code}")
            return []

    except requests.exceptions.RequestException as e:
        print(f"Network error: {e}")
        return []

def get_whit_artwork_details(object_ids):
    """
    Get detailed information about a specific artwork from the Whitney API.

    Args:
        artwork_id: The Whitney Museum's ID for the artwork

    Returns:
        Dictionary with artwork details
    """
    base_url = "https://whitney.org/api"
    url = f"{base_url}/artworks/{object_ids}"

    try:
        response = requests.get(url)

        if response.status_code == 200:
            return response.json()
        else:
            print(f"Could not get details for artwork {object_ids}") # Corrected variable name from artwork_id to object_ids
            return None

    except requests.exceptions.RequestException as e:
        print(f"Network error: {e}")
        return None
search_term = input("Enter a search term for artworks: ")


print(f"\nFetching artworks with '{search_term}' in title...")
artworks_data = search_whit_artworks(title_contains=search_term)
print(f"Artworks with '{search_term}':", artworks_data)

detailed_artworks = []
for artwork_id in artworks_data:
    print(f"Fetching details for artwork ID: {artwork_id}")
    details = get_whit_artwork_details(artwork_id)
    if details:
        detailed_artworks.append(details)
    else:
        print(f"Could not retrieve details for artwork {artwork_id}")
print(f"Successfully fetched details for {len(detailed_artworks)} artworks.")

print("\nSummary of artworks:")
print("-" * 30)
for i, artwork_details in enumerate(detailed_artworks):
    artwork = artwork_details['data']['attributes']
    artist_text = artwork.get('display_artist_text', 'Unknown Artist')
    print(f"Artwork {i+1}:")
    print(f"  Title: {artwork.get('title', 'N/A')}")
    print(f"  Artist: {artist_text}")
    print(f"  Date: {artwork.get('display_date', 'N/A')}")
    print(f"  Medium: {artwork.get('medium', 'N/A')}")
    print(f"  Department: {artwork.get('department', 'N/A')}")
    print(f"  Classification: {artwork.get('classification', 'N/A')}")
    print("-" * 30)

```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from datetime import datetime
import json

def save_report(analysis, filename):
    """Save analysis results to JSON."""
    with open(filename, "w") as file:
        json.dump(analysis, file, indent=2)
    print(f"Report saved to {filename}")

def generate_summary(analysis):
    """Generate human-readable summary."""
    summary = []
    summary.append("=" * 50)
    summary.append("SURVEY ANALYSIS REPORT")
    summary.append("=" * 50)
    summary.append(f"Total Responses: {analysis['total_responses']}")
    summary.append(f"\n {search_term} by DEPARTMENT:")

    for cat, count in analysis["categories"].items():
        percentage = (count / analysis["total_responses"]) * 100
        summary.append(f"  {cat}: {count} ({percentage:.1f}%)")

    summary.append(f"\n{search_term} by CLASSIFICATION:")

    for catClass, count in analysis["categoriesClass"].items():
        percentage = (count / analysis["total_responses"]) * 100
        summary.append(f"  {catClass}: {count} ({percentage:.1f}%)")

    summary.append(f"\n {search_term} by MEDIUM:")

    for catMed, count in analysis["categoriesMed"].items():
        percentage = (count / analysis["total_responses"]) * 100
        summary.append(f"  {catMed}: {count} ({percentage:.1f}%)")

    summary.append(f"\n {search_term.upper()} by COUNTRY:")

    for ctry, count in analysis["country"].items():
        percentage = (count / analysis["total_responses"]) * 100
        summary.append(f"  {ctry}: {count} ({percentage:.1f}%)")

    summary.append(f"\nAnalysis Date: {analysis['analysis_date']}")

    return "\n".join(summary)

# Initialize counters for analysis
department_counts = Counter()
classification_counts = Counter()
medium_counts = Counter()
artist_country_counts = Counter()

for artwork_details in detailed_artworks:
    artwork = artwork_details['data']['attributes']
    department_counts[artwork.get('department', 'Unknown')] += 1
    classification_counts[artwork.get('classification', 'Unknown')] += 1
    medium_counts[artwork.get('medium', 'Unknown')] += 1
    # As artist country is not directly available, using 'Unknown' for this example
    artist_country_counts['Unknown'] += 1

# Create the analysis dictionary
analysis = {
    'total_responses': len(detailed_artworks),
    'categories': dict(department_counts),
    'categoriesClass': dict(classification_counts),
    'categoriesMed': dict(medium_counts),
    'country': dict(artist_country_counts),
    'analysis_date': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}

# Generate the summary report
summary_report = generate_summary(analysis)
print(summary_report)

# The previous plotting code for classifications and departments remains the same
classifications = []
departments = []
medium = [] # Initialize medium list

for artwork_details in detailed_artworks:
    artwork = artwork_details['data']['attributes']
    classifications.append(artwork.get('classification', 'Unknown'))
    departments.append(artwork.get('department', 'Unknown'))
    medium.append(artwork.get('medium', 'Unknown')) # Populate medium list

#creating dataframe
df = pd.DataFrame({
    'Classification': classifications,
    'Department': departments,
    'Medium': medium
})

#plotting Classification
plt.figure(figsize=(12, 6))
df['Classification'].value_counts().plot(kind='bar')
plt.title('Distribution of Artworks by Classification')
plt.xlabel('Classification')
plt.ylabel('Number of Artworks')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

#plotting Department
plt.figure(figsize=(10, 5))
df['Department'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Distribution of Artworks by Department')
plt.xlabel('Department')
plt.ylabel('Number of Artworks')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

#plotting Medium
plt.figure(figsize=(10, 5))
df['Medium'].value_counts().plot(kind='bar', color='lightcoral')
plt.title('Distribution of Artworks by Medium')
plt.xlabel('Medium')
plt.ylabel('Number of Artworks')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
